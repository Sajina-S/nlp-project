```python
para="""hi,hello,welcome, everyone to my utub channel.hope you are doing well.
this is my village.
my heaven."""
```

```python
#load the NLTK module
import nltk

#dwnld punk
nltk.download('punkt')

#diwnld stopwords
nltk.download("stopwords")

#download the wordnet
nltk.download("wordnet")

nltk.download("averaged_perceptron_tagger")
```

```python
from nltk.tokenize import sent_tokenize
```

```python
tokenized_sentences=sent_tokenize(para)
```

```python
print(tokenized_sentences)
len(tokenized_sentences)
```

```python
from nltk.tokenize import word_tokenize
```

```python
tokenized_words=word_tokenize(para)
```

```python
print(tokenized_words)
len(tokenized_words)
```

```python
from nltk.probability import FreqDist
```

```python
fdist=FreqDist(tokenized_words)
print(fdist)
```

```python
print(fdist.most_common(20))
```

```python
import matplotlib.pyplot as plt

fdist.plot(20,cumulative=False)
plt.show()
```

```python
#stop words removal
```

```python
from nltk.corpus import stopwords
stop_words_list=set(stopwords.words('english'))
```

```python
stop_words_list
```

```python
filterwords=[]
for word in tokenized_words:
  if word not in stop_words_list:
    filterwords.append(word)


print(filterwords)
print(tokenized_words)
```

```python
import pandas as pd
```

```python
df=pd.read_csv('/content/amazon_alexa.tsv',sep="\t")
```

```python
df.head()
```

```python
df['feedback'].value_counts()
```

```python
#bag of words done by count vectorisation module
```

```python
df['verified_reviews']
```

```python

from sklearn.feature_extraction.text import CountVectorizer
```

```python
count_vec=CountVectorizer(stop_words='english',lowercase=True) #stopwords removal and making all lowercased
x=count_vec.fit_transform(df['verified_reviews'])
y=df['feedback']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=143)
```

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg=LogisticRegression()
log_reg.fit(x_train,y_train)
y_pred=log_reg.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test,y_pred)
```

```python
print(acc)
```

```python
from sklearn.metrics import confusion_matrix
print('confusion matrix is:\n',confusion_matrix(y_test,y_pred))
```
